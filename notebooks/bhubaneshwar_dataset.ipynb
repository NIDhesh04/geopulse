```python
import osmnx as ox

print("Downloading Bhubaneswar road network...")

G = ox.graph.graph_from_place(
    "Bhubaneswar, Odisha, India",
    network_type="drive",
    simplify=True
)

print("\n========== GRAPH SUMMARY ==========")
print("Nodes:", len(G.nodes))
print("Edges:", len(G.edges))

output_path = "bhubaneswar_drive.graphml"

ox.io.save_graphml(G, output_path)

print("\nSaved graph to:", output_path)
```
